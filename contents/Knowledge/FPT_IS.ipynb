## Data pipeline

🔧 Định nghĩa đơn giản
- Data Pipeline là một hệ thống tự động vận chuyển dữ liệu từ điểm A đến điểm B, đồng thời có thể biến đổi dữ liệu trong quá trình đó.

Hãy tưởng tượng như một đường ống nước:

+ 🚰 Nguồn nước = nơi dữ liệu xuất phát (database, API, file...)
+ 🔧 Đường ống = các bước xử lý, làm sạch, biến đổi
+ 🪣 Bể chứa = nơi dữ liệu đến (Data Warehouse, Dashboard...)

🔄 Luồng hoạt động cơ bản

[Nguồn dữ liệu] → [Thu thập] → [Làm sạch] → [Biến đổi] → [Lưu trữ] → [Sử dụng]

Ví dụ thực tế tại FPT IS (khách hàng Chính phủ):


Hệ thống thuế, hải quan,    →   ETL/ELT   →   Data Warehouse   →   Dashboard

cơ sở dữ liệu dân cư..................................(Oracle, PostgreSQL)..............báo cáo

**📦 Các thành phần của Data Pipeline**

1. **🟢 Data Sources (Nguồn dữ liệu)**

**Dữ liệu có thể đến từ nhiều nơi:**

- Database: Oracle, SQL Server, PostgreSQL
- File: CSV, Excel, JSON, XML
- API: REST API từ các hệ thống khác
- Real-time: Kafka, log hệ thống

2. **🔵 Ingestion (Thu thập)**

- Kéo dữ liệu từ các nguồn vào hệ thống
- Có thể theo batch (theo lô, ví dụ: mỗi đêm 1 lần) hoặc real-time (liên tục)

3. **🟡 Transformation (Biến đổi — ETL/ELT)**

**Đây là bước quan trọng nhất:**

- Làm sạch dữ liệu (xử lý null, duplicate, sai định dạng)
- Chuẩn hóa (đồng nhất đơn vị, kiểu dữ liệu)
- Tính toán, tổng hợp theo nghiệp vụ
- Join nhiều nguồn lại với nhau

4. 🟠 **Storage (Lưu trữ)**

- Data Warehouse (DWH): Lưu dữ liệu đã xử lý để phân tích
- Data Lake: Lưu dữ liệu thô, chưa xử lý
- Data Mart: Tập con của DWH cho từng phòng ban

5. 🔴 **Serving (Phục vụ)**

- Dashboard Power BI / Tableau
- Báo cáo định kỳ
- Input cho ML models

## Data modeling

Data Modeling là gì?

🎯 Định nghĩa đơn giản

Data Modeling là việc thiết kế "bản vẽ" cho dữ liệu — tức là xác định:

- Dữ liệu gồm những bảng nào?
- Mỗi bảng có những cột gì?
- Các bảng liên kết với nhau như thế nào?


Giống như bản vẽ kiến trúc trước khi xây nhà — không có bản vẽ thì xây sẽ loạn!

🏗️ Có 3 loại mô hình chính
1. 🟢 Conceptual Model — Mô hình khái niệm

Mức cao nhất, chỉ mô tả tổng quan, không đi vào chi tiết kỹ thuật.

[Khách hàng] ──── mua ──── [Sản phẩm]

2. 🟡 Logical Model — Mô hình logic

Đi vào chi tiết hơn, xác định các thuộc tính nhưng chưa quan tâm đến database cụ thể.

|KHÁCH HÀNG         | ĐƠN HÀNG                        |
|----------------|-------------------|
| **- MaKH (PK)    1──∞** |  - MaDH (PK) |
| **- HoTen**             | - MaKH (FK) |
| **- Email**             |  - NgayDat |
| **- DiaChi**             | - TongTien |

3. 🔵 Physical Model — Mô hình vật lý

Chi tiết nhất — thiết kế thực tế trong database cụ thể (Oracle, PostgreSQL...)

```
CREATE TABLE KHACH_HANG (
    ma_kh       VARCHAR(10) PRIMARY KEY,
    ho_ten      NVARCHAR(100) NOT NULL,
    email       VARCHAR(100) UNIQUE,
    dia_chi     NVARCHAR(200)
);
```

⭐ Quan trọng nhất trong DWH: 2 Schema
- Đây là phần bạn cần học kỹ nhất vì JD FPT IS yêu cầu DWH + Data Modeling.

📐 Star Schema (Hình sao) — Phổ biến nhất

```
                   [DIM_THOI_GIAN]                    
                   Năm, Quý, Tháng                    
                         |                            
[DIM_KHACH_HANG] ── [FACT_DOANH_THU] ── [DIM_SAN_PHAM]
  Tên, Địa chỉ        DoanhThu          Tên, Danh mục 
                      SoLuong                         
                         |                            
                  [DIM_TINH_THANH]                    
                     Tỉnh, Vùng                       
``` 

Gồm 2 loại bảng:
- FACT table — chứa số liệu đo lường (doanh thu, số lượng, chi phí...)
- DIMENSION table — chứa thông tin mô tả (ai, ở đâu, khi nào, cái gì...)

❄️ Snowflake Schema (Hình bông tuyết)

Giống Star Schema nhưng các Dimension được tách nhỏ hơn:
```
[DIM_VUNG] ← [DIM_TINH_THANH] ← [FACT_DOANH_THU]
```
+ Star Schema = đơn giản, query nhanh → dùng phổ biến hơn
+ Snowflake = tiết kiệm bộ nhớ hơn nhưng query phức tạp hơn

```
Nguồn dữ liệu
      ↓
   ETL/ELT
      ↓
⭐ DATA MODELING ⭐  ← Bạn thiết kế DWH trông như thế nào ở đây
      ↓
  Data Warehouse
  (Star Schema)
      ↓
   Data Marts
      ↓
Power BI / Báo cáo
      
- Data Modeling là bước thiết kế trước khi xây DWH — thiết kế tốt thì query nhanh, báo cáo chuẩn!
```

## Oracle, SQL Server, PostgreSQL

Oracle, SQL Server, PostgreSQL đều là Database Management System (DBMS) — tức là phần mềm quản lý cơ sở dữ liệu.

Chúng có thể đóng nhiều vai trò khác nhau trong pipeline, không chỉ là nơi load dữ liệu về.

Chúng có thể đứng ở BẤT KỲ đâu trong pipeline!

```
📥 NGUỒN DỮ LIỆU         🔄 ETL/ELT         🏢 DWH / Data Mart
┌──────────────┐                            ┌──────────────┐    
│ Oracle       │ ---→  xử lý, làm sạch --→  │ PostgreSQL   │    
│ (hệ thống    │                            │ (kho tổng    │    
│  nghiệp vụ)  │                            │  hợp)        │    
└──────────────┘                            └──────────────┘    
┌──────────────┐                                                
│ SQL Server   │  --→  xử lý, làm sạch --→  │ Oracle       │    
│ (hệ thống    │                            │ (DWH chính)  │    
│  khác)       │                            └──────────────┘    
└──────────────┘                                                
```
Cùng 1 phần mềm nhưng vai trò khác nhau tùy vào vị trí trong pipeline!

Phân biệt cho dễ nhớ

| Công cụ | Oracle / SQL Server / PostgreSQL |
|---------|----------------------------------|
| **Bản chất** | Phần mềm quản lý database   |
| **Vai trò 1** | Nguồn dữ liệu (hệ thống nghiệp vụ đang chạy) |
| **Vai trò 2** | Nơi lưu DWH sau ETL/ELT ✅ |
| **Vai trò 3** | Lưu Data Mart cho từng phòng ban |

Vậy tại sao JD yêu cầu biết cả 3?

Vì ở FPT IS, khách hàng Chính phủ có thể dùng hệ thống khác nhau:

Bộ Tài Chính  → đang dùng Oracle

Bộ Công An    → đang dùng SQL Server  

Dự án mới     → dùng PostgreSQL (mã nguồn mở, miễn phí)


```
Oracle / SQL Server / PostgreSQL               
           ↓                                   
    Chỉ là "thùng chứa"                        
    ┌─────────────────┐                        
    │  Lưu trữ data   │  ← Có thể là NGUỒN     
    │  (đọc & ghi)    │  ← Có thể là ĐÍCH (DWH)
    └─────────────────┘                        
    ❌ KHÔNG xử lý/làm sạch                    
```
